Sequential Prompt Chaining Flow:

START --> Topic --> LLM1 --> Blog Outline --> LLM2 --> Detailed Blog based on Outline --> Response --> END

```python
from langgraph.graph import StateGraph,START,END
from typing import TypedDict
from langchain_google_genai import ChatGoogleGenerativeAI
from dotenv import load_dotenv
load_dotenv()
```

```python
class LLMState(TypedDict):
    topic: str
    blog_outline: str
    detailed_blog: str
```

```python
def LLM_blog_outline(state:LLMState)->LLMState:
    model1 = ChatGoogleGenerativeAI(model="gemini-2.5-pro")

    prompt = f"Generate a detailed outline for blog on topic {state['topic']}"

    blog_outline = model1.invoke(prompt).content

    state['blog_outline'] = blog_outline

    return state
```

```python
def LLM_detailed_blog(state:LLMState)->LLMState:
    model2 = ChatGoogleGenerativeAI(model="gemini-2.5-pro")

    prompt = f"Generate a detailed blog on topic:{state['topic']} based on the blog outline:{state['blog_outline']} provided"

    detailed_blog = model2.invoke(prompt).content
    
    state['detailed_blog'] = detailed_blog

    return state
```

```python
graph = StateGraph(LLMState)
```

```python
graph.add_node('LLM_blog_outline',LLM_blog_outline)
graph.add_node('LLM_detailed_blog',LLM_detailed_blog)
```

```python
graph.add_edge(START,'LLM_blog_outline')
graph.add_edge('LLM_blog_outline','LLM_detailed_blog')
graph.add_edge('LLM_detailed_blog',END)
```

```python
workflow = graph.compile()
```

```python
input = {'topic':"The Rise of AI in INDIA"}

output = workflow.invoke(input)

print("===========================================================")
print("================ BLOG OUTLINE =============================")
print(output['blog_outline'])
print("===========================================================")
```

```python
print("===========================================================")
print("======================= BLOG  =============================")
print(output['detailed_blog'])
print("===========================================================")
```
