Sequential Prompt Chaining Flow 2:

START --> Topic --> LLM1 --> Blog Subject --> LLM2 --> Blog 200 Words --> Response --> END

```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langgraph.graph import StateGraph, START,END
from typing import TypedDict
from dotenv import load_dotenv
load_dotenv()
```

```python
class LLM1State(TypedDict):
    query : str
    subject: str
    blog : str
```

```python
def subject(state: LLM1State)-> LLM1State:
    model1 = ChatGoogleGenerativeAI(model="gemini-2.5-pro")
    response = model1.invoke(f"Generate a blog subject for the topic : {state['query']}").content
    state['subject']=response
    return state
```

```python
def blog(state: LLM1State)-> LLM1State:
    model2 = ChatGoogleGenerativeAI(model="gemini-2.5-pro")
    response = model2.invoke(f"Generate a blog of 200 words for the topic : {state['query']} and subject :{state['subject']}").content
    state['blog']=response
    return state
```

```python
graph = StateGraph(LLM1State)
```

```python
graph.add_node("Blog_Subject",subject)
graph.add_node("Blog",blog)
```

```python
graph.add_edge(START,"Blog_Subject")
graph.add_edge("Blog_Subject","Blog")
graph.add_edge("Blog",END)
```

```python
workflow = graph.compile()
```

```python
input = {'query':"AI in Tech"}

output = workflow.invoke(input)
```
